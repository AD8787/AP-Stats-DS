```python
#SETTING UP THE REST OF THE CODE–THIS CELL IS MY GENERAL SETUP FOR THE BLENDLAB REPORT

#Import all necessary libraries

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

url = "https://raw.githubusercontent.com/mikechechelnitsky-cpu/APStatsDS27/main/BlendLab_Transactions.xlsx"

df = pd.read_excel(url)

df.head()
```


# ***PART 1:***

```python
#Getting the length of the DataFrame df gives us the total number of transactions because it returns the total nunmber of rows in the dataframe
#since the rows are based on transaction ID through the column "Transaction_ID," the total number of rows is the number of transactions

total_transactions = len(df)
print(f"Total Transactions: {total_transactions}", "\n")

#setting mean and medians for price, customer age, and wait time.
mean_price = df["Price"].mean()
median_price = df["Price"].median()

mean_age = df["Customer_Age"].mean()
median_age = df["Customer_Age"].median()

mean_wait_time = df["Wait_Time_Min"].mean()
median_wait_time = df["Wait_Time_Min"].median()

#Appropriate measures of spread for 2 variables, I choose price and customer age and get the standard deviation and iqr using built in functions
#For price:
stdev_price = df["Price"].std()

Q3_price = df["Price"].quantile(0.75)
Q1_price = df["Price"].quantile(0.25)
iqr_price = Q3_price - Q1_price

#For customer age:
stdev_age = df["Customer_Age"].std()

Q3_age = df["Customer_Age"].quantile(0.75)
Q1_age = df["Customer_Age"].quantile(0.25)
iqr_age = Q3_age - Q1_age

#Printing all data for price variable
print("Mean, Median, & Measures of Spread for Product Price: ")
print(f"- Mean of Product Price: ${mean_price:.2f}")
print(f"- Median of Product Price: ${median_price:.2f}")
print(f"- Standard Deviation of Product Price: ${stdev_price:.2f}")
print(f"- IQR of Product Price: ${iqr_price:.2f}", "\n")

#Printing all data and measures of spread for customer age variable
print("Mean, Median, & Measures of Spread for Customer Age: ")
print(f"- Mean of Customer Age: {mean_age:.2f}")
print(f"- Median of Customer Age: {median_age:.2f}")
print(f"- Standard Deviation of Customer Age: {stdev_age:.2f}")
print("- IQR of Customer Age:", iqr_age, "\n")

#Printing mean and median for wait time
print("Mean & Median for Wait Time: ")
print(f"- Mean of Wait Time: {mean_wait_time:.2f}")
print(f"- Median of Wait Time: {median_wait_time:.2f}\n")

#Five number summary, I chose the variable wait time, and I renamed some of the row values
summary_wait_time = df["Wait_Time_Min"].describe().loc[["min", "25%", "50%", "75%", "max"]].rename({"25%": "Q1", "50%": "median", "75%": "Q3"})

print("5 Number Summary for Wait Time: ")
print(summary_wait_time)
```

### **Explanation for Pt. 1:**

- **Mean:** The mean tells the average of whatever data value we are trying to find. With the calculations, we can find that the average price for a BlendLab product is $9.54, and the average age of a customer buying something from BlendLab is 32.54 years old. The average time that they have to wait in line to recieve their order is 5.08 minutes

- **Median:** The median tells us the middle value of the data set column we are looking for. In the case of BlendLabs, we find the middle price value for each of the data points, including the median of product price ($9.50), customer age (31 years old), wait time (4.50 mins), which are the values that are located directly in the middle of their respective data columns when they are sorted in order from least to greatest.

- **Min (e.g. for wait time):** This gives us the minimum value for wait time, which we found that the lowest amount of time that someone had to wait for their order is 1.6 minutes.

- **Max (e.g. for wait time):** This gives us the maximum value for wait time, which we found that the highest amount of time that someone had to wait for their order is 20.3 minutes.

- **Q1 (e.g. for wait time):** Q1 represents the 25th percentile of the dataset. It is the point where 25% of the data falls below it, while 75% of the data falls above it. Essentially, it is the median of the lower half of the data. In context of wait time, Q1 is 3.3, which means that the median amount of time people had to wait when they were in the 25th percentile of wait time is 3.3 minutes.

- **Q3 (e.g. wait time):** Q3 represents the 75th percentile of the dataset. It is the point where 75% of the data falls below it, while 25% of the data falls above it. Essentially, it is the median of the upper half of the data. In context of wait time, Q3 is 6.0, which means that the median amount of time people had to wait when they were in the 75th percentile of wait time is 6.0 minutes.




# ***PART 2:***

```python
Q1_wait = df["Wait_Time_Min"].quantile(0.25)
Q3_wait = df["Wait_Time_Min"].quantile(0.75)
IQR_wait = Q3_wait - Q1_wait
upper_bound_wait = Q3_wait + (1.5 * IQR_wait)
lower_bound_wait = Q1_wait - (1.5 * IQR_wait)

outliers_wait = df[
    (df["Wait_Time_Min"] > upper_bound_wait)
    | (df["Wait_Time_Min"] < lower_bound_wait)
]
print(f"Wait Time IQR: {IQR_wait:.2f} | Upper Bound: {upper_bound_wait:.2f} min")
print(f"Number of potential wait time outliers: {len(outliers_wait)}")

# Visualization 1 & 2: Histogram for Price and Boxplot for Wait Time (Matplotlib)
fig, axes = plt.subplots(1, 2, figsize = (14, 5))

axes[0].hist(df["Price"], bins=15, color="skyblue", edgecolor="black")
axes[0].set_title("Distribution of Transaction Prices")
axes[0].set_xlabel("Price ($)")
axes[0].set_ylabel("Frequency")

axes[1].boxplot(df["Wait_Time_Min"], vert = True, patch_artist=True, boxprops = dict(facecolor="lightgreen"))
axes[1].set_title("Boxplot of Customer Wait Times")
axes[1].set_ylabel("Wait Time (Minutes)")
axes[1].set_xticks([1])
axes[1].set_xticklabels(["Wait Time"])

plt.tight_layout()
plt.show()
```
