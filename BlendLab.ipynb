```python
#SETTING UP THE REST OF THE CODE–THIS CELL IS MY GENERAL SETUP FOR THE BLENDLAB REPORT

#Import all necessary libraries

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

url = "https://raw.githubusercontent.com/mikechechelnitsky-cpu/APStatsDS27/main/BlendLab_Transactions.xlsx"

df = pd.read_excel(url)

df.head()
```

```python
#get frequency of each product in dataset
product_counts = df["Product"].value_counts()
product_counts.plot(kind="pie")
plt.show()
```


# ***PART 1:***

```python
#Getting the length of the DataFrame df gives us the total number of transactions because it returns the total nunmber of rows in the dataframe
#since the rows are based on transaction ID through the column "Transaction_ID," the total number of rows is the number of transactions

total_transactions = len(df)
print(f"Total Transactions: {total_transactions}", "\n")

#setting mean and medians for price, customer age, and wait time.
mean_price = df["Price"].mean()
median_price = df["Price"].median()

mean_age = df["Customer_Age"].mean()
median_age = df["Customer_Age"].median()

mean_wait_time = df["Wait_Time_Min"].mean()
median_wait_time = df["Wait_Time_Min"].median()

#Appropriate measures of spread for 2 variables, I choose price and customer age and get the standard deviation and iqr using built in functions
#For price:
stdev_price = df["Price"].std()

Q3_price = df["Price"].quantile(0.75)
Q1_price = df["Price"].quantile(0.25)
iqr_price = Q3_price - Q1_price

#For customer age:
stdev_age = df["Customer_Age"].std()

Q3_age = df["Customer_Age"].quantile(0.75)
Q1_age = df["Customer_Age"].quantile(0.25)
iqr_age = Q3_age - Q1_age

#Printing all data for price variable
print("Mean, Median, & Measures of Spread for Product Price: ")
print(f"- Mean of Product Price: ${mean_price:.2f}")
print(f"- Median of Product Price: ${median_price:.2f}")
print(f"- Standard Deviation of Product Price: ${stdev_price:.2f}")
print(f"- IQR of Product Price: ${iqr_price:.2f}", "\n")

#Printing all data and measures of spread for customer age variable
print("Mean, Median, & Measures of Spread for Customer Age: ")
print(f"- Mean of Customer Age: {mean_age:.2f}")
print(f"- Median of Customer Age: {median_age:.2f}")
print(f"- Standard Deviation of Customer Age: {stdev_age:.2f}")
print("- IQR of Customer Age:", iqr_age, "\n")

#Printing mean and median for wait time
print("Mean & Median for Wait Time: ")
print(f"- Mean of Wait Time: {mean_wait_time:.2f}")
print(f"- Median of Wait Time: {median_wait_time:.2f}\n")

#Five number summary, I chose the variable wait time, and I renamed some of the row values
summary_wait_time = df["Wait_Time_Min"].describe().loc[["min", "25%", "50%", "75%", "max"]].rename({"25%": "Q1", "50%": "median", "75%": "Q3"})

print("5 Number Summary for Wait Time: ")
print(summary_wait_time)
```

### **Explanation for Pt. 1:**

- **Mean:** The mean tells the average of whatever data value we are trying to find. With the calculations, we can find that the average price for a BlendLab product is $9.54, and the average age of a customer buying something from BlendLab is 32.54 years old. The average time that they have to wait in line to recieve their order is 5.08 minutes

- **Median:** The median tells us the middle value of the data set column we are looking for. In the case of BlendLabs, we find the middle price value for each of the data points, including the median of product price ($9.50), customer age (31 years old), wait time (4.50 mins), which are the values that are located directly in the middle of their respective data columns when they are sorted in order from least to greatest.

- **Min (e.g. for wait time):** This gives us the minimum value for wait time, which we found that the lowest amount of time that someone had to wait for their order is 1.6 minutes.

- **Max (e.g. for wait time):** This gives us the maximum value for wait time, which we found that the highest amount of time that someone had to wait for their order is 20.3 minutes.

- **Q1 (e.g. for wait time):** Q1 represents the 25th percentile of the dataset. It is the point where 25% of the data falls below it, while 75% of the data falls above it. Essentially, it is the median of the lower half of the data. In context of wait time, Q1 is 3.3, which means that the median amount of time people had to wait when they were in the 25th percentile of wait time is 3.3 minutes.

- **Q3 (e.g. wait time):** Q3 represents the 75th percentile of the dataset. It is the point where 75% of the data falls below it, while 25% of the data falls above it. Essentially, it is the median of the upper half of the data. In context of wait time, Q3 is 6.0, which means that the median amount of time people had to wait when they were in the 75th percentile of wait time is 6.0 minutes.




# ***PART 2:***

```python
#for wait time
Q1_wait = df["Wait_Time_Min"].quantile(0.25)
Q3_wait = df["Wait_Time_Min"].quantile(0.75)

IQR_wait = Q3_wait - Q1_wait
upper_bound_wait = Q3_wait + (1.5 * IQR_wait) #calculates IQR upper bound
lower_bound_wait = Q1_wait - (1.5 * IQR_wait) #calculates IQR lower bound

print(df["Wait_Time_Min"].describe().loc[["std", "25%", "75%"]]) #prints only the std, 1st quartile, and 3rd quartile

outliers_wait = [] #creates list for outliers of wait time

for value in df["Wait_Time_Min"]: #iterates through all the values in wait time column
  if value < lower_bound_wait: #if the wait time is less than the lower bound, add that wait time into outliers_wait
    outliers_wait.append(value)

  elif value > upper_bound_wait: #if the wait time is more than the lower bound, add that wait time into outliers_wait
    outliers_wait.append(value)

print(outliers_wait) #print the list of outliers_wait
print("There are " + str(len(outliers_wait)) + " wait time outliers_wait.") #prints the number of outliers_wait

#histogram for wait time
print("\n")
plt.figure(figsize = (8,4))
plt.hist(df["Wait_Time_Min"], bins = 25, edgecolor = "black", linewidth = 0.5, color = "blue")
plt.title("Distribution of Wait Times")
plt.xlabel("Wait Time (Minutes)")
plt.ylabel("Frequency")
plt.show()

# for customer age
Q1_age = df["Customer_Age"].quantile(0.25)
Q3_age = df["Customer_Age"].quantile(0.75)

IQR_age = Q3_age - Q1_age
upper_bound_age = Q3_age + (1.5 * IQR_age)  # calculates IQR upper bound
lower_bound_age = Q1_age - (1.5 * IQR_age)  # calculates IQR lower bound

print("\n")
print(df["Customer_Age"].describe().loc[["std", "25%", "75%"]])  # prints only the std, 1st quartile, and 3rd quartile

outliers_age = []  # creates list for outliers of customer age

for value in df["Customer_Age"]:  # iterates through all the values in customer age column
  if value < lower_bound_age:
    outliers_age.append(value)  # if the age is less than the lower bound, add it into outliers_age

  elif value > upper_bound_age:
    outliers_age.append(value)  # if the age is more than the upper bound, add it into outliers_age

print(outliers_age)  # print the list of outliers_age
print("There are " + str(len(outliers_age)) + " customer age outliers.")  # prints the number of outliers_age

# histogram for customer age
print("\n")
plt.figure(figsize = (8, 4))
plt.hist(df["Customer_Age"], bins = 25, edgecolor = "black", linewidth = 0.5, color = "green")
plt.title("Distribution of Customer Ages")
plt.xlabel("Customer Age (Years)")
plt.ylabel("Frequency")
plt.show()
```

### **Explanation for Pt. 2:**

- **Wait Time:** For the wait times, those high values stretching out near 20 minutes are most likely not errors, since they could take into account when the shop could be in a rush or some other issues that happened that pushed back orders (such as a machine malfunction preventing the barista from making smoothies).

- **Customer Age:** As for the customer ages, the older people up around 70 years old are totally reasonable since older people buy smoothies and açaí bowls too. They are not considered errors as the outlier represents customers years and there is no restriction to when someone is able to order here.

# ***PART 4:***

```python
#1st Comparison: customer types by Marketing Campaign

#Social Media
social_media = df[df["Marketing_Campaign"] == "Social Media"]
print("Social Media Customer Types: ")
print(social_media["Customer_Type"].value_counts(), "\n")

#Digital Ads
digital_ads = df[df["Marketing_Campaign"] == "Digital Ads"]
print("Digital Ads Customer Types: ")
print(digital_ads["Customer_Type"].value_counts(), "\n")

#Email/Loyalty
email_loyalty = df[df["Marketing_Campaign"] == "Email/Loyalty"]
print("Email/Loyalty Customer Types: ")
print(email_loyalty["Customer_Type"].value_counts(), "\n")

#Local Partnership
local_partner = df[df["Marketing_Campaign"] == "Local Partnership"]
print("Local Partnership Customer Types: ")
print(local_partner["Customer_Type"].value_counts(), "\n")

#No Active Campaign
no_campaign = df[df["Marketing_Campaign"] == "No Active Campaign"]
print("No Active Campaign Customer Types: ")
print(no_campaign["Customer_Type"].value_counts(), "\n")


#2nd comparison: Discount Usage by Marketing Campaign

print("\n")

# Let's see how many people used a discount for each campaign using a simple loop
campaigns = ["Social Media", "Digital Ads", "Email/Loyalty", "Local Partnership", "No Active Campaign"]

print("Discount Usage Breakdown by Campaign: \n")
for camp in campaigns:
  # Filter the dataframe for each campaign
  subset = df[df["Marketing_Campaign"] == camp]
  # Count Yes/No for discounts
  discounts = subset["Discount_Used"].value_counts()
  print(f"Campaign: {camp}")
  print(discounts, "\n")

# Visualization: Simple Bar Chart for Discounts
# This creates a crosstab table: https://pandas.pydata.org/docs/reference/api/pandas.crosstab.html$0

discount_table = pd.crosstab(df["Marketing_Campaign"], df["Discount_Used"])

plt.figure(figsize=(10, 5))
discount_table.plot(kind="bar", color=["tomato", "limegreen"])
plt.title("Discount Usage Across Different Marketing Campaigns")
plt.xlabel("Marketing Campaign")
plt.ylabel("Number of Transactions")
plt.xticks(rotation = 25)
plt.legend(title = "Discount Used")
plt.grid(True, linestyle = "--", alpha = 0.5)
plt.show()
```

### **Explanation for Pt. 4:**

- When looking at the data and breaking it down by type of campaign, the Email/Loyalty and the Digital ad programs have a much higher retenton rate of returning customers. In contrast, the social media, local partnership, and no active campaigns had less returning customers, and this makes sense because email/loyalty programs already tend to have loyal customers (hence the name) and digital ads help spread word on the internet. However, the type of product depends as well, as some products (like we saw earlier that have a higher likelihood of being bought depending on location) could have higher retention rates in those specific areas, which pushes people to buy based on product, not just based on whether they like BlendLab as a whole or not

- Looking at the discount usage, the email/loyalty and social media campaigns use more discounts, and they also have the most number of transactions. This also makes sense because of the promotion a brand could have on social media, and email/loyalty customers likely having points or other systems that grant them discounts. It also lets us know that other customers are willing to pay more full price, albiet to the fact that they may not buy at all in the first place.



# ***PART 5:***

Do Discounts increase or decrease customer satisfaction ratings?

```python
import matplotlib.pyplot as plt
import pandas as pd

# Count how many Yes/No discounts exist for each satisfaction rating
counts = pd.crosstab(df["Satisfaction_Rating"], df["Discount_Used"])

# Keep columns in the desired order ("No" then "Yes") using pandas.DataFrame.reindex:
#https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.reindex.html$0

counts = counts.reindex(columns=["No", "Yes"], fill_value = 0)

print(counts, "\n")

# Plot a bar chart
counts.plot(kind = "bar", figsize = (10, 5), color = ["maroon", "green"])

plt.title("Discount Usage by Satisfaction Rating")
plt.xlabel("Satisfaction Rating")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.legend(title = "Discount Used")
plt.tight_layout()
plt.show()
```

### **Explanation for Pt. 5:**

- From the conclusion from the graph above, we can see that when the discount is not used, the customers are actually more likely to have a higher satisfaction rating, like a 4 or a 5 star review.
- However, customers that have a disount are more likely to leave a 4 or 5 star review than leave a lower review, like a 3 star.
- In summary, offering a discount does not automatically boost customer satisfaction to a 5 star rating. In fact, customers without discounts have more 5 star ratings than 4 star ratings, while they are mostly evenly split for 4 and 5 star for customers with discounts
